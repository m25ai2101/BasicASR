```python
from gtts import gTTS
from IPython.display import Audio
import re

# Step 1: Input text
text = "Hello everyone. Today I will demonstrate how text to speech systems work."

print("Original Text:")
print(text)

# Step 2: Basic text preprocessing
clean_text = re.sub(r'[^a-zA-Z0-9 .,!?]', '', text)

print("\nCleaned Text:")
print(clean_text)

# Step 3: Convert text to speech
tts = gTTS(text=clean_text, lang="en")

# Step 4: Save audio
file_path = "speech_demo.mp3"
tts.save(file_path)

print("\nSpeech generated!")

# Step 5: Play audio
Audio(file_path)
```

```python
from gtts import gTTS
from IPython.display import Audio
import re

# Step 1: Take user input
text = input("Enter text to convert to speech: ")

# Step 2: Basic preprocessing
clean_text = re.sub(r'[^a-zA-Z0-9 .,!?]', '', text)

print("\nProcessed text:")
print(clean_text)

# Step 3: Converting the i/p text from user to speech
tts = gTTS(text=clean_text, lang="en")

# Step 4: Save audio
file_path = "generated_speech.mp3"
tts.save(file_path)

print("\nSpeech generated successfully!")

# Step 5: Play audio
Audio(file_path)
```
