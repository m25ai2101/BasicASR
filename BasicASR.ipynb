```python
import whisper

print("Loading model...")
model = whisper.load_model("base")
print("Model loaded!")
```

```python
from IPython.display import Javascript, display
from google.colab import output
from base64 import b64decode
import time

def record_audio_manual(filename="input.wav"):
    print("🎤 Click START → Speak → Click STOP")

    js_code = """
    async function record() {
        const div = document.createElement('div');
        const start = document.createElement('button');
        const stop = document.createElement('button');

        start.textContent = 'Start Recording';
        stop.textContent = 'Stop Recording';

        div.appendChild(start);
        div.appendChild(stop);
        document.body.appendChild(div);

        let recorder;
        let chunks = [];

        start.onclick = async () => {
            const stream = await navigator.mediaDevices.getUserMedia({ audio: true });
            recorder = new MediaRecorder(stream);
            recorder.ondataavailable = e => chunks.push(e.data);
            recorder.start();
        };

        return new Promise(resolve => {
            stop.onclick = () => {
                recorder.stop();
                recorder.onstop = async () => {
                    const blob = new Blob(chunks);
                    const reader = new FileReader();
                    reader.readAsDataURL(blob);
                    reader.onloadend = () => {
                        const base64data = reader.result.split(',')[1];
                        div.remove();  // ✅ REMOVE BUTTONS
                        resolve(base64data);
                    };
                };
            };
        });
    }

    record();
    """

    audio_data = output.eval_js(js_code)

    with open(filename, "wb") as f:
        f.write(b64decode(audio_data))

    print("✅ Recording saved!")
    return filename
```

```python
audio_file = record_audio_manual()

print("🧠 Transcribing...")
result = model.transcribe(audio_file, fp16=False)

print("\n📝 Transcription:")
print(result["text"])
```
